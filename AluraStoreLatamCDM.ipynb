##Ventas Totales vs Productos vendidos por Tienda

```python
df_graf_ventasXTienda = pd.DataFrame({
    'Tiendas': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'TiendaVentasTot': [TiendaTotal, Tienda2Total, Tienda3Total, Tienda4Total]
})
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

# ============================================================
# 1. UNIR TODAS LAS TIENDAS EN UN SOLO DATAFRAME
# ============================================================

tienda['Tienda'] = 'tienda'
tienda2['Tienda'] = 'tienda2'
tienda3['Tienda'] = 'tienda3'
tienda4['Tienda'] = 'tienda4'

df = pd.concat([tienda, tienda2, tienda3, tienda4], ignore_index=True)


# ============================================================
# 2. CALCULAR TOTALES POR TIENDA
# ============================================================

productos_por_tienda = (
    df.groupby('Tienda')
      .size()
      .reset_index(name='TotalProductosVendidos')
)

ventas_por_tienda = (
    df.groupby('Tienda')['Precio']
      .sum()
      .reset_index(name='TotalVentas')
)

df_totales = productos_por_tienda.merge(ventas_por_tienda, on='Tienda')


# ============================================================
# 3. GRÁFICO COMBINADO CON EJE SECUNDARIO
# ============================================================

plt.figure(figsize=(12,7))

x = range(len(df_totales))

# Eje principal (productos vendidos)
ax1 = plt.gca()
ax1.bar(
    x,
    df_totales['TotalProductosVendidos'],
    color='skyblue',
    width=0.4,
    label='Productos Vendidos'
)

# Etiquetas de productos vendidos
for i, row in df_totales.iterrows():
    ax1.text(
        i,
        row['TotalProductosVendidos'] + 5,
        f"{row['TotalProductosVendidos']:,}".replace(",", "."),
        ha='center',
        fontsize=11,
        weight='bold'
    )

ax1.set_ylabel('Productos Vendidos')
ax1.set_xlabel('Tienda')
ax1.set_xticks(x)
ax1.set_xticklabels(df_totales['Tienda'])
ax1.grid(axis='y', linestyle='--', alpha=0.4)

# Formato del eje Y (productos vendidos)
ax1.yaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x, _: f"{int(x):,}".replace(",", "."))
)


# Eje secundario (ventas totales)
ax2 = ax1.twinx()
ax2.plot(
    x,
    df_totales['TotalVentas'] / 1e6,
    color='seagreen',
    marker='o',
    linewidth=3,
    label='Ventas Totales (Millones)'
)

# Etiquetas de ventas totales (más arriba para mejor lectura)
for i, row in df_totales.iterrows():
    ax2.text(
        i,
        (row['TotalVentas'] / 1e6) - 6,   # ← etiqueta más arriba
        f"{row['TotalVentas']/1e6:,.f}".replace(",", "."),
        ha='center',
        fontsize=11,
        color='seagreen',
        weight='bold'
    )

ax2.set_ylabel('Ventas Totales (Millones)')

# Formato del eje Y (ventas)
ax2.yaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x, _: f"{x:,.0f}".replace(",", "."))
)


# ============================================================
# 4. TÍTULO Y LEYENDA
# ============================================================

plt.title('Productos Vendidos y Ventas Totales por Tienda')

# Leyenda en la esquina inferior izquierda
lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
plt.legend(
    lines_1 + lines_2,
    labels_1 + labels_2,
    loc='lower left'
)

plt.show()
```

##Calificacion vs Ventas por tienda

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

# ============================================================
# 1. UNIR TODAS LAS TIENDAS EN UN SOLO DATAFRAME
# ============================================================

# Agregar columna con el nombre de la tienda
tienda['Tienda'] = 'tienda'
tienda2['Tienda'] = 'tienda2'
tienda3['Tienda'] = 'tienda3'
tienda4['Tienda'] = 'tienda4'

# Unir todo en un solo DataFrame maestro
df = pd.concat([tienda, tienda2, tienda3, tienda4], ignore_index=True)


# ============================================================
# 2. CALCULAR CATEGORÍAS DISTINTAS VENDIDAS POR TIENDA (OPCIÓN A)
# ============================================================

categorias_por_tienda = (
    df.groupby('Tienda')['Categoría del Producto']
      .nunique()
      .reset_index()
)

categorias_por_tienda.columns = ['Tienda', 'CategoriasVendidas']


# ============================================================
# 3. CALCULAR VENTAS, CALIFICACIÓN Y ENVÍO PROMEDIO POR TIENDA
# ============================================================

df_graf_ventasXcalificacionXenvio = (
    df.groupby('Tienda')
      .agg({
          'Precio': 'sum',
          'Calificación': 'mean',
          'Costo de envío': 'mean'
      })
      .reset_index()
)

df_graf_ventasXcalificacionXenvio.columns = [
    'Tienda', 'VentasTotales', 'Calificacion', 'Envio'
]


# ============================================================
# 4. UNIR CATEGORÍAS VENDIDAS AL DATAFRAME FINAL
# ============================================================

df_graf_ventasXcalificacionXenvio = df_graf_ventasXcalificacionXenvio.merge(
    categorias_por_tienda,
    on='Tienda',
    how='left'
)


# ============================================================
# 5. MATRIZ DE OPORTUNIDAD (Scatter con 4 dimensiones)
# ============================================================

plt.figure(figsize=(12,8))

scatter = plt.scatter(
    df_graf_ventasXcalificacionXenvio['VentasTotales'] / 1e6,   # Ventas en millones
    df_graf_ventasXcalificacionXenvio['Calificacion'],          # Calificación
    c=df_graf_ventasXcalificacionXenvio['Envio'],               # Color = costo de envío
    s=df_graf_ventasXcalificacionXenvio['CategoriasVendidas'] * 100,  # Tamaño = categorías
    cmap='viridis',
    edgecolors='black',
    alpha=0.9
)

# ============================================================
# 6. ETIQUETAS DE CADA TIENDA
# ============================================================

for i, row in df_graf_ventasXcalificacionXenvio.iterrows():
    plt.text(
        row['VentasTotales'] / 1e6,
        row['Calificacion'],
        f"{row['CategoriasVendidas']} cat",
        ha='center',
        va='center',
        fontsize=10,
        color='white',
        weight='bold'
    )



# ============================================================
# 7. FORMATO DEL EJE X (separador de miles)
# ============================================================

plt.gca().xaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x, _: f"{int(x):,}".replace(",", "."))
)


# ============================================================
# 8. BARRA DE COLOR (también con separador de miles)
# ============================================================

cbar = plt.colorbar(scatter)
cbar.set_label('Costo de Envío Promedio')
cbar.ax.yaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x, _: f"{int(x):,}".replace(",", "."))
)


# ============================================================
# 9. ESTÉTICA GENERAL
# ============================================================

plt.xlabel('Ventas Totales (Millones)')
plt.ylabel('Calificación Promedio')
plt.title('Matriz de Oportunidad: Ventas vs Calificación (Color = Precio Envío, Tamaño = N°Categorías)')

for spine in plt.gca().spines.values():
    spine.set_visible(False)

plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```
